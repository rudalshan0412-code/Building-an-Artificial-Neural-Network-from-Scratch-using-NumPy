```python
def AND(a, b):
  w1, w2, theta = 0.5, 0.5, 0.7
  #이때 theta의 값은 tmp의 값이 0, 0.5, 1.0이기에 0.7 고정은 아니다
  tmp = a*w1 + b*w2
  if tmp <= theta:
    return 0
  elif tmp > theta:
    return 1
#AND 게이트
```

```python
#이걸 numpy로 구현하면

import numpy as np

def AND(x1, x2):

  x = np.array([x1, x2]) #배열 [x1, x2]
  w = np.array([0.5, 0.5])

  b = -0.7
# w*x = array([0.5*x1, 0.5*x2])
  np.sum(w*x)
#0.5*x1 + 0.5*x2
  tmp = np.sum(w*x) + b
  if tmp <= 0:
    return 0
  else:
    return 1
```

```python
def NAND(x1, x2):

  x = np.array([x1, x2]) #배열 [x1, x2]
  w = np.array([-0.5, -0.5])

  b = 0.7
# w*x = array([0.5*x1, 0.5*x2])
  np.sum(w*x)
#0.5*x1 + 0.5*x2
  tmp = np.sum(w*x) + b
  if tmp <= 0:
    return 0
  else:
    return 1
```

```python
def OR(x1, x2):

  x = np.array([x1, x2]) #배열 [x1, x2]
  w = np.array([0.5, 0.5])

  b = -0.2
# w*x = array([0.5*x1, 0.5*x2])
  np.sum(w*x)
#0.5*x1 + 0.5*x2
  tmp = np.sum(w*x) + b
  if tmp <= 0:
    return 0
  else:
    return 1

#OR 게이트이다. 이때, 가중치가 없어도 np.sum의 값은 0 0 일때를 제외하면 전부 양수
```

```python
# AND 와 NAND, OR 게이트들은 x * w + b의 형태로 값 * 가중치 + 편향의 형태로 나타낼 수 있다.

# 퍼셉트론은 평면 위의 점을 직선 하나로 나누는 장치이다.

# 이때, XOR은 직선 하나로 나눌 수 없다.

# 따라서, 하나의 퍼셉트론으로는 XOR을 구현할 수 없다

#이때, 단순 if를 사용하는것은 논리적으로는 구현이 가능하나, 사람이 직접 추가한 것이지

# 퍼셉트론이 직접 학습한 것이 아니다.

# 이를 해결하기 위해서, 2개의 퍼셉트론을 사용하면 된다
```

```python
def XOR(x1, x2):

  s1 = OR(x1, x2)
  s2 = NAND(x1, x2)

  y = AND(s1, s2)

  return y

#이러면 XOR을 정의할 수 있다.
```

```python
#의의 XOR은 가중치를 지정해준 것임
#XOR의 가중치를 랜덤으로 시작해서 정의 및 최적화 시키기
```

```python
def sigmoid(x):
  return 1/(1 + np.exp(-x)) # np.exp는 자연상수 e의 제곱값을 반환한다

# 어떠한 값이 나와도 0~1 사이의 값이 나온다. 이때, x의 값이 커질수록 반환값이 1에 가까워진다
```

```python
# XOR은 단층으로 구현핤 수 없기에 입력 1개(x = np. array([x1, x2])), 은닉층 2개, 출력층 1개로 구현한다.
# 이때 가중치는 w1(입력충 -> 은닉층), w2(은닉층 -> 출력층), b1(입력층 -> 은닉층 bias), b2(은닉층 -> 출력층 bias)이 존재한다
```

```python
# 랜덤한 가중치와 x값
np.random.seed(0)

w1 = np.random.randn(2, 2)
b1 = np.zeros((1, 2))

w2 = np.random.randn(2, 1)
b2 = np.zeros((1, 1))

x = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
```

```python
#은닉층
a1 = np.dot(x, w1) + b1 # 이때 가중치는 정의되지 않은, 일종의 랜덤한 가중치이다
#이때 w1.shape = (2, 2)이다. 왜냐하면 x가 원소가 2개인 배열 4개를 담고 있는 2차원 배열이기 때문이다
z1 = sigmoid(a1)
#출력층
a2 = np.dot(z1, w2) + b2
#w2.shape = (2, 1)이다
y = sigmoid(a2)
```

```python
y_pred = y # 랜덤한 가중치를 입력했을때의 예측값
y_true = np.array([[0], [1], [1], [0]]) # x값에 대한 XOR 값
loss = np.mean((y_pred - y_true) ** 2) #초기 loss값
```

```python
#손실함수 구하기
loss = np.mean((y_pred - y_true) ** 2) # 예측값이 정답에 가까울수록 loss가 작아진다(0에 가까워진다)/정답에서 멀어질수록 loss가 커진다(1에 가까워진다)
# print(loss)
```

```python
#역전파 구하기

#순서는 손실함수 -> y_pred -> a2 -> w2, b2 -> a1 -> w1, b1 순으로 진행한다
```

```python
#손실함수
d_loss = 2 * (y_pred - y_true) / len(y_true)
#d_loss의 값이 0에 가까울수록 답에 근접한다
#loss 값을 미분하여 평균을 구하는 것이기에 2를 곱하고 평균값을 구하기에 len(y_true)로 나눈다
#d_loss값이 음수인 경우 더 키워야 답에 근접하고, 양수인 경우 낮춰야 답에 근접한다
```

```python
#y_pred

y_pred_derivative = y_pred*(1 - y_pred)
#sigmoid 함수의 미분은 f'(x) = f(x)(1 - f(x))이다

d_a2 = d_loss * y_pred_derivative #a2 값에 따른 출력값의 변화량
# d_a2.shape = (4, 1)
```

```python
#w2, b2

d_w2 = np.dot(z1.T, d_a2)
# w2.shape = (2, 1) 이기에 d_w2.shape = (2, 1)이다
# a2 = z1_1 * w2_1 + z1_2 * w2_2 + b2 이다.
# 이 때, w2의 변화량에 따른 loss의 변화량은 (w2의 변화량에 따른 a2의 변화량)*d_a2이다.

#d_w2 = ([d_w2_1], [d_w2_2])
#d_w2_1 = 은닉뉴런1의 출력 * 출력층 오차 전부의 합(은닉뉴런 1이 출력층에 주는 영향의 크기)
#d_w2_2 = 은닉뉴런2의 출력 * 출력층 오차 전부의 합(은닉뉴런 2이 출력층에 주는 영향의 크기)
#이를 한번에 계산하기 위해 z1.T = [[데이터1의 은닉1, 데이터 2의 은닉1 ...], [데이터 1의 은닉 2, 데이터 2의 은닉 2, ...]]

d_b2 = np.sum(d_a2, axis = 0, keepdims = True)
#b2는 a2를 구할 때 곱해지는 것이 아니기에 오차 d_a2를 전부 더하면 된다
#axis = 0은 세로방향으로 되어있는 da2를 전부 세로방향으로 더해준다
# keepdims = True 는 d_b2와 b2의 shape를 맞춰준다
```

```python
#z1

d_z1 = np.dot(d_a2, w2.T)
# a2 = z1_1 * w2_1 + z1_2 * w2_2 + b2이다. 이때 z1에 대해서 미분하면 w2남게 된다.
# 이때, d_z1은 z1이 loss에 주는 영향을 주는지 계산하는 것이기에
# z1의 미분 * d_a2 를 하게 된다

#z1 = sigmoid(a1) 이기에 sidmoid 함수의 미분을 곱해주어야한다

d_a1 = d_z1 * z1 * (1 - z1)
```

```python
#w1, b1

d_w1 = np.dot(x.T, d_a1)

d_b1 = np.sum(d_a1, axis = 0, keepdims = True)
```

```python
#가중치 업데이트

learning_rate = 1.0
# 학습률(한번의 학습 결과가 얼마나 영향을 줄지 결정)

w1 -= learning_rate * d_w1
b1 -= learning_rate * d_b1

w2 -= learning_rate * d_w2
b2 -= learning_rate * d_b2
```

```python
# 전체 코드

import numpy as np


def sigmoid(x):
  return 1/(1 + np.exp(-x))

# x값과 랜덤한 가중치

x = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])

np.random.seed(0)

w1 = np.random.randn(2, 2)
b1 = np.zeros((1, 2))

w2 = np.random.randn(2, 1)
b2 = np.zeros((1, 1))

#랜덤값 확인
print(f"초기 w1:{w1}\n 초기 b1:{b1}]\n 초기 w2:{w2}\n 초기 b2:{b2}")

learning_rate = 1.0

for epoch in range(10000):
  a1 = np.dot(x, w1) + b1
  z1 = sigmoid(a1)
  a2 = np.dot(z1, w2) + b2
  y = sigmoid(a2)

  y_pred = y
  y_true = np.array([[0], [1], [1], [0]])
  loss = np.mean((y_pred - y_true) ** 2)

  d_loss = 2 * (y_pred - y_true) / len(y_true)


  y_pred_derivative = y_pred*(1 - y_pred)

  d_a2 = d_loss * y_pred_derivative

  d_w2 = np.dot(z1.T, d_a2)

  d_b2 = np.sum(d_a2, axis = 0, keepdims = True)

  d_z1 = np.dot(d_a2, w2.T)

  d_a1 = d_z1 * z1 * (1 - z1)

  d_w1 = np.dot(x.T, d_a1)

  d_b1 = np.sum(d_a1, axis = 0, keepdims = True)

  #가중치 업데이트

  w1 -= learning_rate * d_w1
  b1 -= learning_rate * d_b1

  w2 -= learning_rate * d_w2
  b2 -= learning_rate * d_b2

  #중간 확인
  if epoch % 1000 == 0:
    print(epoch, loss)

print(f"loss값 :{loss} \n 예측값:{y_pred}\n w1:{w1}\n b1:{b1}\n w2:{w2}\n b2:{b2}")
```

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# x값과 랜덤한 가중치

x = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
np.random.seed(0)

w1 = np.random.randn(2, 2)
b1 = np.zeros((1, 2))
w2 = np.random.randn(2, 1)
b2 = np.zeros((1, 1))


#랜덤값 확인
print(f"초기 w1:{w1}\n 초기 b1:{b1}]\n 초기 w2:{w2}\n 초기 b2:{b2}")

learning_rate = 1.0
epoch = 0

# 무한루프 대비
max_epochs = 100000

while epoch < max_epochs:

    a1 = np.dot(x, w1) + b1
    z1 = sigmoid(a1)
    a2 = np.dot(z1, w2) + b2
    y_pred = sigmoid(a2)

    y_true = np.array([[0], [1], [1], [0]])

    loss = np.mean((y_pred - y_true) ** 2)

    # loss가 0.1 이하로 떨어지는 순간 break
    if loss < 0.1:
        break

    d_loss = 2 * (y_pred - y_true) / len(y_true)
    y_pred_derivative = y_pred * (1 - y_pred)
    d_a2 = d_loss * y_pred_derivative

    d_w2 = np.dot(z1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_z1 = np.dot(d_a2, w2.T)
    d_a1 = d_z1 * z1 * (1 - z1)

    d_w1 = np.dot(x.T, d_a1)
    b1_derivative = np.sum(d_a1, axis=0, keepdims=True)

    # 가중치 업데이트
    w1 -= learning_rate * d_w1
    b1 -= learning_rate * b1_derivative
    w2 -= learning_rate * d_w2
    b2 -= learning_rate * d_b2

    # 진행상황 확인
    if epoch % 5000 == 0:
        print(f"Epoch {epoch:5d} | Current Loss: {loss:.6f}")

    epoch += 1
if epoch != 100000:
  print(f"목표 달성 epoch 값: {epoch}")
  print(f"최종 loss 값: {loss:.6f}")
  print(f"최종 예측값:\n{y_pred}")

else:
  print("목표 달성하지 못함")
```
